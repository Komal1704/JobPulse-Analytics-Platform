```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine(
    "mysql+pymysql://root:123456@localhost:3307/jobpulse"
)
```

```python
df_salary = pd.read_csv(
    r"C:\Projects\JobPulse\data\cleaned\cleaned_salaries.csv"
)

df_salary.head()
```

```python
df_salary.to_sql(
    name="salaries",
    con=engine,
    if_exists="replace",
    index=False
)

print("Salary dataset uploaded successfully!")
```

```python
df_jobs = pd.read_csv(
    r"C:\Projects\JobPulse\data\cleaned\cleaned_job_postings.csv"
)

df_jobs.head()
```

```python
df_jobs.to_sql(
    name="job_postings",
    con=engine,
    if_exists="replace",
    index=False
)

print("Job postings uploaded successfully!")
```
