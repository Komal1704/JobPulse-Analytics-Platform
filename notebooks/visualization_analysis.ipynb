```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_salary = pd.read_csv(
    r"C:\Projects\JobPulse\data\cleaned\cleaned_salaries.csv"
)

df_jobs = pd.read_csv(
    r"C:\Projects\JobPulse\data\cleaned\cleaned_job_postings.csv"
)
```

**Highest Paying Roles**

```python
df_salary.groupby("job_title")[
    "salary_in_usd"
].mean().sort_values(ascending=False).head(10).plot(kind="bar")

plt.title("Top 10 Highest Paying Roles")
plt.ylabel("Average Salary")
plt.show()
```

**Work Type Distribution**

```python
df_jobs["formatted_work_type"].value_counts().plot(kind="bar")

plt.title("Job Work Type Distribution")
plt.ylabel("Total Jobs")
plt.show()
```

**Experience Level Demand**

```python
df_jobs["formatted_experience_level"].value_counts().plot(kind="bar")

plt.title("Experience Level Demand")
plt.ylabel("Total Jobs")
plt.show()
```

**Remote Work Distribution**

```python
df_salary["remote_ratio"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Remote Work Distribution")
plt.ylabel("")
plt.show()
```

**Salary Transparency**

```python
df_jobs["salary_disclosed"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Salary Disclosure Distribution")
plt.ylabel("")
plt.show()
```
